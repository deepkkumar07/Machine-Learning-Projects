```python
# importing basic/required libraries
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import KFold, cross_val_score, StratifiedKFold
from sklearn.metrics import accuracy_score, mean_squared_error
```

```python
import seaborn as sns
df = sns.load_dataset('iris')
df.head()
```

```python
df['species'] = df['species'].map({'setosa':0, 'versicolor':1, 'virginica':2})
```

```python
df
```

```python
X= df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
y = df['species']
```

#K-Fold Cross-Validation

**Standard K-Fold Cross-Validation** is a data-splitting technique primarily used for Regression tasks (predicting continuous numbers like housing prices, temperatures, or salaries) where there are no distinct categorical classes to balance.

```python
clf = LogisticRegression(max_iter=1000)
k =5
score = cross_val_score(clf,X,y, cv=k)
score
```

```python
average_score = np.mean(score)
average_score
```

# *startified k fold*

**Stratified K-Fold** is a specialized version of K-Fold cross-validation designed specifically for **Classification** problems.

 While regular KFold splits the data randomly, StratifiedKFold ensures that each fold contains roughly the same percentage of each class as the entire dataset.

```python
clf = LogisticRegression(max_iter=1000)
stf = StratifiedKFold(n_splits=5)
score1 = cross_val_score(clf,X,y, cv=stf)
score1
```

```python
print('avg score', np.mean(score1))
```

### Checking for Underfitting and Overfitting

To assess whether the model is underfitting or overfitting, we'll train the `LogisticRegression` classifier on the entire dataset and evaluate its accuracy on the training data. We will then compare this training accuracy to the average cross-validation score previously calculated.

```python
clf_full_data = LogisticRegression(max_iter=1000)
clf_full_data.fit(X, y)

training_accuracy = clf_full_data.score(X, y)
print(f'Training Accuracy: {training_accuracy:.4f}')
print(f'Average Cross-Validation Accuracy: {average_score:.4f}')

if training_accuracy > average_score + 0.05: # A small threshold for comparison
    print('The model might be overfitting: Training accuracy is significantly higher than cross-validation accuracy.')
elif training_accuracy < average_score - 0.05:
    print('The model might be underfitting: Training accuracy is significantly lower than cross-validation accuracy.')
else:
    print('The model appears to have a good balance between bias and variance.')
```
