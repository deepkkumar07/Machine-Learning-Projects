```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

```python
df = sns.load_dataset('iris')
```

```python
x = df[['sepal_length','sepal_width','petal_length','petal_width']].copy()
y = df['species'].copy()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# model building
model = LogisticRegression(multi_class='multinomial',solver= 'lbfgs', max_iter= 200)
# multi_class'multinomial'Instructs the model to use Multinomial Logistic Regression (also known as Softmax Regression). Instead of training multiple independent binary classifiers, it directly models the true probability distribution across all categories simultaneously, ensuring the probabilities for all classes sum up to 1.
#solver'lbfgs'Specifies the optimization algorithm used to find the best mathematical weights (coefficients) to minimize error.
#max_iter200Sets the maximum number of iterations the solver is allowed to take to converge on the best solution. The default in scikit-learn is 100, so raising it to 200 gives the L-BFGS algorithm more time to find the optimal weights if the dataset is complex or difficult to separate.
```

```python
#training model
model.fit(x_train, y_train)
```

```python
#prediction
y_pred= model.predict(x_test)
```

```python
#Evaluate
print(classification_report(y_test, y_pred))
```
